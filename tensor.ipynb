```python
import tensorflow as tf
a=tf.constant([[3,4,5,4],[4,5,6,3],[2,3,5,4]])
print(a)
print(a[1])
print(a[0])
print(a[2,2])
print(a[1,0])
print(a[2,1])
b=tf.transpose(a)
print(b)
print(a[1,0])
print(a[2:4])
print(a[1:,2:])
print(a[0:2,1:3])
print(a[-1:-2:-1,-3:-4:-1])
print(a[-1:-2:-1])
e=tf.constant([[2,3,4,5],[2,6,8,4],[3,6,9,1]])
x=tf.reverse(e,axis=[1,0])
print(x)
```

```python
import tensorflow as tf
a=tf.constant([2,4,5,6]) 
print(tf.reshape(a,(2,2)))
print(tf.reshape(a,(-1,2)))
print(tf.reshape(a,(1,-1)))
b=tf.constant([[2,3,4],[3,6,8],[5,6,8]])
print(tf.reshape(b,[-1]))
c=tf.constant([1,2,3])
d=tf.constant([3,4,5])
print(tf.add(c,d))
print(tf.subtract(c,d))
print(tf.multiply(c,d))
print(tf.divide(c,d))
```

```python
import tensorflow as tf
a=tf.constant([[2,3],[4,6],[6,8]])
b=tf.constant([[7,8],[3,9]])
c=tf.matmul(a,b)
print(c)
```

```python
import tensorflow as tf
a=tf.constant([[3,4],[5,6],[7,8]])
c=tf.reduce_mean(a)
print(c)
d=tf.reduce_sum(a)
print(d)
print(tf.reduce_max(a))
print("\n",tf.reduce_min(a))
print(tf.reduce_sum(a,axis=0))
print(tf.reduce_sum(a,axis=1))
print(tf.reduce_max(a,axis=1))
```
